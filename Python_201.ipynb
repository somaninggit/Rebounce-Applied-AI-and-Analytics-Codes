Track A: Personal Expense Tracker

This text-based tool helps you monitor your daily financial spending by logging expenses with their categories, descriptions, and amounts.
It provides analytical views of your data, allowing you to instantly see your total spending and a breakdown of costs by category.
All data is securely saved to and loaded from a JSON file, ensuring your financial history is preserved between sessions.
Features included:

1.   Add a new expense: Prompts for category, amount, and description with  error handling for invalid numbers.
2.   View all expenses: Displays a clean, readable list of everything you have logged.
3.  View total amount spent: Calculates the grand total of all your expenses.
4.  View totals by category: Groups your spending to show how much you spent on specific categories (e.g., Food, Transport).
5.  Save to file: Automatically saves all your data to a JSON file.
Load from file: Reads the JSON file when the program starts so no past data is lost.
Interactive Menu: A menu that keeps running in a loop until you specifically choose to "save and quit".







```python
import json
import os
```

```python
# 1. Load data from file
def load_expenses(filename="expenses.json"):
    """Loads expenses from a JSON file. Returns an empty list if file doesn't exist."""
    if os.path.exists(filename):
        try:
            with open(filename, 'r') as file:
                return json.load(file)
        except json.JSONDecodeError:
            print("Error reading the file. Starting with empty expenses.")
            return []
    return []
```

```python
# 2. Save data to file
def save_expenses(expenses, filename="expenses.json"):
    """Saves the current list of expenses to a JSON file."""
    with open(filename, 'w') as file:
        json.dump(expenses, file, indent=4)
    print("Expenses saved successfully.")
```

```python
# 3. Add a new expense
def add_expense(expenses):
    """Prompts the user for category, description, and amount with error handling."""
    category = input("Category: ").strip().title()
    description = input("Description: ").strip()

    # Graceful error handling for input validation
    while True:
        try:
            amount = float(input("Amount: "))
            if amount < 0:
                print("Amount cannot be negative. Please try again.")
                continue
            break
        except ValueError:
            print("Invalid input! Please enter a valid number for the amount.")

    expense = {
        "category": category,
        "amount": amount,
        "description": description
    }
    expenses.append(expense)
    print("Added.")
```

```python
# 4. View all expenses
def view_all_expenses(expenses):
    """Displays all logged expenses in a readable format."""
    if not expenses:
        print("No expenses recorded yet.")
        return

    print("\n--- All Expenses ---")
    for idx, exp in enumerate(expenses, 1):
        print(f"{idx}. {exp['category']} - Rs {exp['amount']:.2f} ({exp['description']})")
    print("--------------------")
```

```python

# 5. View totals by category
def view_totals(expenses):
    """Calculates and prints the total spending per category and the grand total."""
    if not expenses:
        print("No expenses to summarize.")
        return

    totals = {}
    grand_total = 0.0

    for exp in expenses:
        cat = exp['category']
        amt = exp['amount']
        totals[cat] = totals.get(cat, 0.0) + amt
        grand_total += amt

    print("\n--- Totals by Category ---")
    for cat, total in totals.items():
        print(f"{cat}: Rs {total:.2f}")
    print(f"Total: Rs {grand_total:.2f}")
    print("--------------------------")
```

```python

# 6. Main Orchestration
def main():
    """Main menu loop tying all functions together."""
    expenses = load_expenses()

    while True:
        print("\n=== Expense Tracker ===")
        print("1. Add expense")
        print("2. View all")
        print("3. View totals by category")
        print("4. Save and quit")

        choice = input("Choice: ").strip()

        if choice == '1':
            add_expense(expenses)
        elif choice == '2':
            view_all_expenses(expenses)
        elif choice == '3':
            view_totals(expenses)
        elif choice == '4':
            save_expenses(expenses)
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1, 2, 3, or 4.")

# Entry point
if __name__ == "__main__":
    main()
```
