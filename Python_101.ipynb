```python
password = "hello@2026"

length_ok = len(password) >= 8
contains_at = "@" in password
strong = length_ok and contains_at

print(f"Password: {password}")
print(f"Length OK (>= 8): {length_ok}")
print(f"Contains '@': {contains_at}")
print(f"Strong: {strong}")
```

```python
scores = {
    "Anika": 35,
    "Rohan": 72,
    "Meera": 40,
    "Kabir": 28,
    "Diya": 88,
    "Ishaan": 55
}

passing_students = list(
    map(
        lambda student: student[0],
        filter(lambda student: student[1] >= 40, scores.items())
    )
)

print("Passing students:", passing_students)
```
